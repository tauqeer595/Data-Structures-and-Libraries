```python
import numpy as np
import pandas as pd

data = {'Values': [10, 12, 14, 15, 18, 210, 19, 17, 16, 14]}
df = pd.DataFrame(data)

# Calculate mean and standard deviation
mean = np.mean(df['Values'])
std_dev = np.std(df['Values'])

# Compute Z-scores
df['Z_Score'] = (df['Values'] - mean) / std_dev

# Identify outliers (Z-score > 3 or < -3)
outliers = df[np.abs(df['Z_Score']) > 2]

print("Dataset with Z-scores:\n", df)
print("\nOutliers detected:\n", outliers)
```

**What is a Box Plot?**

A box plot is used to represent the distribution of numerical data. It shows median, quartiles, and outliers. It's great for visualizing the spread and central tendency of the data.

The box represents the interquartile range (IQR), i.e., the middle 50% of the data.

The line inside the box represents the median (50th percentile).

Whiskers show the range of the data.

Outliers are values that fall outside the whiskers.

```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt

# Load a sample dataset (Iris dataset is a good one)
data = sns.load_dataset('iris')

# Create a box plot
sns.boxplot(x='species', y='sepal_length', data=data)

# Show the plot
plt.title("Box Plot for Sepal Length by Species")
plt.show()
```

```python
import dash
from dash import dcc, html
import plotly.express as px
from jupyter_dash import JupyterDash

# Load data
df = px.data.gapminder().query("country == 'India'")

# Line Chart
fig1 = px.line(df, x='year', y='gdpPercap', title='GDP Per Capita')

# Bar Chart
fig2 = px.bar(df, x='year', y='pop', title='Population')

# Initialize App
app = JupyterDash(__name__)

# Layout with Multiple Charts
app.layout = html.Div([
    html.H1('Dashboard with Multiple Charts'),
    dcc.Graph(figure=fig1),
    dcc.Graph(figure=fig2)
])

# Run the App inside Colab
app.run(mode='inline')
```

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
from jupyter_dash import JupyterDash

# Load data
df = px.data.gapminder()

# Initialize App
app = JupyterDash(__name__)

# App Layout with Dropdown
app.layout = html.Div([
    html.H1('GDP Per Capita by Country'),

    dcc.Dropdown(
        id='country-dropdown',
        options=[{'label': c, 'value': c} for c in df['country'].unique()],
        value='India'
    ),

    dcc.Graph(id='line-graph')
])

# Callback for Interactivity
@app.callback(
    Output('line-graph', 'figure'),
    Input('country-dropdown', 'value')
)
def update_graph(selected_country):
    filtered_df = df[df['country'] == selected_country]
    fig = px.line(filtered_df, x='year', y='gdpPercap', title=f'GDP Per Capita of {selected_country}')
    return fig

# Run in Colab
app.run(mode='inline')
```

```python
import dash
from dash import dcc, html, Input, Output, dash_table
import plotly.express as px
from jupyter_dash import JupyterDash
import pandas as pd

# Load Data
df = px.data.gapminder()

# Initialize App
app = JupyterDash(__name__)

# App Layout
app.layout = html.Div([
    html.H1('🌍 World Data Dashboard', style={'textAlign': 'center', 'color': 'blue'}),

    # Tabs
    dcc.Tabs([

        dcc.Tab(label='Interactive Charts', children=[

            html.Div([
                html.Label('Select Continent:'),
                dcc.Dropdown(id='continent-dropdown',
                             options=[{'label': i, 'value': i} for i in df['continent'].unique()],
                             value='Asia'),

                html.Label('Select Country:'),
                dcc.Dropdown(id='country-dropdown'),

                html.Label('Select Chart Type:'),
                dcc.RadioItems(
                    id='chart-type',
                    options=[
                        {'label': 'Line Chart', 'value': 'line'},
                        {'label': 'Bar Chart', 'value': 'bar'}
                    ],
                    value='line',
                    labelStyle={'display': 'inline-block', 'margin-right': '10px'}
                ),

                html.Label('Select Year Range:'),
                dcc.RangeSlider(
                    id='year-slider',
                    min=df['year'].min(),
                    max=df['year'].max(),
                    step=5,
                    marks={str(year): str(year) for year in df['year'].unique()},
                    value=[1952, 2007]
                ),

                dcc.Graph(id='main-graph')
            ])
        ]),

        dcc.Tab(label='Data Table', children=[
            dash_table.DataTable(
                id='data-table',
                columns=[{"name": i, "id": i} for i in df.columns],
                page_size=10,
                filter_action='native',
                sort_action='native',
                style_table={'overflowX': 'auto'},
                style_header={'backgroundColor': 'lightblue', 'fontWeight': 'bold'},
                style_cell={'textAlign': 'left', 'padding': '5px'}
            )
        ])
    ])
])

# Callback for Dynamic Country Dropdown
@app.callback(
    Output('country-dropdown', 'options'),
    Input('continent-dropdown', 'value')
)
def set_country_options(selected_continent):
    return [{'label': i, 'value': i} for i in df[df['continent'] == selected_continent]['country'].unique()]

# Callback for Charts
@app.callback(
    Output('main-graph', 'figure'),
    [Input('country-dropdown', 'value'),
     Input('chart-type', 'value'),
     Input('year-slider', 'value')]
)
def update_graph(selected_country, chart_type, year_range):
    filtered_df = df[(df['country'] == selected_country) &
                     (df['year'] >= year_range[0]) &
                     (df['year'] <= year_range[1])]

    if chart_type == 'line':
        fig = px.line(filtered_df, x='year', y='gdpPercap', color='country',
                      title=f'GDP Per Capita - {selected_country}')
    else:
        fig = px.bar(filtered_df, x='year', y='gdpPercap', color='country',
                     title=f'GDP Per Capita - {selected_country}')

    fig.update_layout(template='plotly_dark')
    return fig

# Data Table Callback
@app.callback(
    Output('data-table', 'data'),
    Input('continent-dropdown', 'value')
)
def update_table(selected_continent):
    return df[df['continent'] == selected_continent].to_dict('records')

# Run in Colab
app.run(mode='inline')
```
